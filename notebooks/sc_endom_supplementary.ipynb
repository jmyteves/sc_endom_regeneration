# sc_endom supplementary figures

Supplementary figures S1-S5 of the same manuscript, plus the panels that were built for earlier versions and are not in the current figure set.

```python
import os
import sys


def find_repo_root():
    """Walk up from the working directory until src/, data/ and scripts/ are all present."""
    if os.environ.get("SC_ENDOM_ROOT"):
        return os.path.abspath(os.environ["SC_ENDOM_ROOT"])
    d = os.path.abspath(os.getcwd())
    while True:
        if all(os.path.isdir(os.path.join(d, k)) for k in ("src", "data", "scripts")):
            return d
        parent = os.path.dirname(d)
        if parent == d:
            raise RuntimeError("repository root not found; set SC_ENDOM_ROOT")
        d = parent


REPO_ROOT = find_repo_root()
FIGS = os.path.join(REPO_ROOT, "figures")
os.makedirs(FIGS, exist_ok=True)
sys.path.insert(0, os.path.join(REPO_ROOT, "src"))
print("repository:", REPO_ROOT)
print("figures   :", FIGS)
```

## Check

Every SVG must carry live text in a named font.

```python
import glob

import render_svg

svgs = sorted(glob.glob(os.path.join(FIGS, "SFig*.svg")))
bad = []
for p in svgs:
    v = render_svg.verify(p)
    if v["text_elements"] == 0 or not v["named_font"]:
        bad.append(v)
print(len(svgs), "SVG files;", len(bad), "without live text")
for v in bad:
    print(v)
```
