# sc_endom main figures

Mouse uterus scRNA-seq time course (Control, E16.5, PPD1, PPD7, PPD30; 698,631 cells) with histology, morphometry and organoid assays. Each cell below rebuilds one main figure from `data/` into `figures/`. Nothing here opens the atlas h5ad; `data/staged/` already holds everything derived from it.

```python
import os
import sys


def find_repo_root():
    """Walk up from the working directory until src/, data/ and scripts/ are all present."""
    if os.environ.get("SC_ENDOM_ROOT"):
        return os.path.abspath(os.environ["SC_ENDOM_ROOT"])
    d = os.path.abspath(os.getcwd())
    while True:
        if all(os.path.isdir(os.path.join(d, k)) for k in ("src", "data", "scripts")):
            return d
        parent = os.path.dirname(d)
        if parent == d:
            raise RuntimeError("repository root not found; set SC_ENDOM_ROOT")
        d = parent


REPO_ROOT = find_repo_root()
FIGS = os.path.join(REPO_ROOT, "figures")
os.makedirs(FIGS, exist_ok=True)
sys.path.insert(0, os.path.join(REPO_ROOT, "src"))
print("repository:", REPO_ROOT)
print("figures   :", FIGS)
```

## Figure 1. Gross involution and epithelial continuity

A timeline, B horn length, C Tukey matrix, F luminal epithelium : stroma, G endometrium / uterus area, H glands per cross-section. D and E are microscopy and return a placeholder.

Code: `src/fig1_panels.py`

```python
import fig1_panels

out = fig1_panels.build(REPO_ROOT, FIGS)
for label, path in out.items():
    if isinstance(path, str) and os.path.exists(path):
        print(f"{label:>10}  {os.path.relpath(path, REPO_ROOT)}")
    elif isinstance(path, str):
        print(f"{label:>10}  {path}")
```

## Check

Every SVG must carry live text in a named font.

```python
import glob

import render_svg

svgs = sorted(glob.glob(os.path.join(FIGS, "Fig*.svg")))
bad = []
for p in svgs:
    v = render_svg.verify(p)
    if v["text_elements"] == 0 or not v["named_font"]:
        bad.append(v)
print(len(svgs), "SVG files;", len(bad), "without live text")
for v in bad:
    print(v)
```
